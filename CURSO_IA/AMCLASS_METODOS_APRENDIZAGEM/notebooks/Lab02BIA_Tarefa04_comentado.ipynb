## Comparação entre Regressão Linear e Regressão Logística

Este notebook carrega a base `leiloes.csv`, prepara as variáveis de entrada e a variável-alvo `Class`, divide os dados em treinamento e teste e compara dois modelos.

> **Observação importante:** como `Class` representa uma classe, a **Regressão Logística** é o modelo mais apropriado. A Regressão Linear é incluída apenas como comparação didática, pois produz valores contínuos e não foi feita para classificação.

## 1. Importação das bibliotecas

As bibliotecas abaixo são usadas para manipulação de dados, separação entre treinamento e teste, criação dos modelos e, futuramente, gráficos.

```python
# pandas: leitura e manipulação de tabelas de dados.
import pandas as pd

# numpy: operações numéricas e vetores/matrizes.
import numpy as np

# Função para separar os dados em conjuntos de treinamento e teste.
from sklearn.model_selection import train_test_split

# Módulo que reúne modelos de regressão linear e logística.
from sklearn import linear_model

# Biblioteca para criação de gráficos.
# Ela foi importada no código original, embora não seja usada nesta tarefa.
import matplotlib.pyplot as plt
```

## 2. Leitura da base de dados

Defina o caminho do arquivo `leiloes.csv`. No VSCode, é mais seguro colocar o CSV na mesma pasta do notebook e utilizar somente o nome do arquivo.

```python
# Caminho do arquivo CSV.
# Caso o arquivo esteja na mesma pasta deste notebook, mantenha apenas "leiloes.csv".
# O caminho original era: "C:/Users/abalb/Downloads/leiloes.csv"
CAMINHO_DADOS = "leiloes.csv"

# Lê o arquivo CSV e armazena o resultado em um DataFrame chamado dados.
dados = pd.read_csv(CAMINHO_DADOS)
```

## 3. Visualização inicial

`head()` mostra as cinco primeiras linhas. É uma verificação rápida para confirmar que o arquivo foi carregado corretamente e para observar as colunas disponíveis.

```python
# Exibe as cinco primeiras linhas da tabela.
dados.head()
```

## 4. Remoção de identificadores

As colunas `Record_ID`, `Auction_ID` e `Bidder_ID` são identificadores. Em geral, identificadores não descrevem o comportamento do fenômeno e podem introduzir padrões artificiais no modelo, por isso são removidos.

```python
# Remove três colunas identificadoras.
# axis=1 indica que a operação ocorre nas colunas.
# inplace=True altera diretamente o DataFrame dados, sem criar uma cópia.
dados.drop(["Record_ID", "Auction_ID", "Bidder_ID"], axis=1, inplace=True)
```

## 5. Conferência após a remoção

Executamos `head()` novamente para verificar se as colunas foram realmente excluídas.

```python
# Mostra novamente as cinco primeiras linhas após a remoção das colunas.
dados.head()
```

## 6. Separação entre entradas (X) e saída (y)

- `X`: contém as variáveis explicativas usadas pelos modelos.
- `y`: contém a variável-alvo `Class`, que o modelo deve prever.

A divisão 70/30 reserva 70% dos dados para treinamento e 30% para teste. `stratify=y` preserva aproximadamente a proporção das classes nos dois conjuntos. `random_state=1` torna a divisão reprodutível.

```python
# X recebe todas as colunas, exceto a coluna-alvo Class.
X = dados.drop("Class", axis=1)

# y recebe somente a coluna-alvo que será prevista.
y = dados["Class"]

# Divide os dados em treinamento e teste.
# test_size=0.3: 30% das observações vão para teste.
# random_state=1: fixa a aleatoriedade para reproduzir a mesma divisão.
# stratify=y: mantém a distribuição das classes em treino e teste.
X_trein, X_teste, y_trein, y_teste = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=1,
    stratify=y
)
```

## 7. Tamanho dos conjuntos

O atributo `shape` informa `(número de linhas, número de colunas)`. Esta célula permite confirmar se a divisão foi realizada como esperado.

```python
# Exibe as dimensões dos conjuntos de treinamento e teste.
print(
    "Tamanho do Conjunto de Treinamento: {}, "
    "Tamanho do Conjunto de Teste: {}".format(
        X_trein.shape,
        X_teste.shape
    )
)
```

## 8. Regressão Linear

A regressão linear busca estimar uma saída numérica contínua. Ela não é a escolha conceitualmente correta quando `Class` é categórica, mas é executada aqui para comparar seu comportamento com o da regressão logística.

```python
# Cria o objeto do modelo de Regressão Linear.
linear = linear_model.LinearRegression()

# Ajusta o modelo usando as variáveis de entrada e a saída do conjunto de treinamento.
linear.fit(X_trein, y_trein)
```

## 9. Predições da Regressão Linear

Como esse modelo gera valores contínuos, as saídas podem não corresponder exatamente às classes válidas. Por isso, não se deve interpretar essas predições diretamente como rótulos de classificação.

```python
# Gera previsões para o conjunto de teste e mostra as 10 primeiras.
linear.predict(X_teste)[:10]
```

## 10. Valores reais correspondentes

Esta célula mostra os 10 primeiros valores reais de `Class`, permitindo a comparação com as previsões anteriores.

```python
# Mostra os 10 primeiros valores reais da variável-alvo no conjunto de teste.
y_teste[:10].values
```

## 11. Escores da Regressão Linear

O método `score()` da regressão linear retorna o coeficiente de determinação R². Esse indicador é próprio para problemas de regressão, não para classificação.

```python
# Calcula e apresenta o R² nos conjuntos de treinamento e teste.
print(
    "Escore do Conjunto de Treinamento: {}, "
    "Escore do Conjunto de Teste: {}".format(
        linear.score(X_trein, y_trein),
        linear.score(X_teste, y_teste)
    )
)
```

## 12. Regressão Logística

A regressão logística é apropriada para classificação. Ela estima a probabilidade de cada observação pertencer a uma classe e, em seguida, atribui a classe mais provável.

```python
# Cria o modelo de Regressão Logística.
logit = linear_model.LogisticRegression(max_iter=1000)

# Ajusta o modelo aos dados de treinamento.
# max_iter=1000 aumenta o limite de iterações e ajuda a evitar aviso de não convergência.
logit.fit(X_trein, y_trein)
```

## 13. Predições da Regressão Logística

Agora as previsões são rótulos de classe, portanto podem ser comparadas diretamente com `y_teste`.

```python
# Prevê as classes do conjunto de teste e exibe as 10 primeiras.
logit.predict(X_teste)[:10]
```

## 14. Valores reais para comparação

Os valores abaixo correspondem às mesmas observações previstas pela regressão logística.

```python
# Exibe os 10 primeiros rótulos reais do conjunto de teste.
y_teste[:10].values
```

## 15. Avaliação da Regressão Logística

Para classificadores, `score()` retorna a **acurácia**, isto é, a proporção de classificações corretas. Quanto mais próximo de 1, melhor o desempenho no conjunto avaliado.

```python
# Calcula e apresenta a acurácia em treinamento e teste.
print(
    "Acurácia do Conjunto de Treinamento: {}, "
    "Acurácia do Conjunto de Teste: {}".format(
        logit.score(X_trein, y_trein),
        logit.score(X_teste, y_teste)
    )
)
```

## Conclusão

Nesta tarefa, o fluxo de classificação foi: carregar os dados, remover identificadores, separar `X` e `y`, dividir em treinamento/teste, treinar modelos e comparar seus resultados. Para prever `Class`, deve-se usar como resultado principal a **Regressão Logística** e sua acurácia no conjunto de teste.
