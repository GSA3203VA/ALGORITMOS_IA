# Objetivo:
# prever a variável "label" usando dados clínicos das pacientes.
# ============================================================


```python
# Importa a biblioteca pandas.
# O pandas é usado para ler e organizar dados em tabelas, chamadas de DataFrame.
import pandas as pd

# Lê o arquivo CSV e armazena os dados na variável "base".
# ATENÇÃO:
# Este caminho funciona somente no computador em que o arquivo está localizado nesta pasta específica.
# Para usar no VSCode, é melhor colocar o CSV dentro da pasta do projeto e usar, por exemplo:
base = pd.read_csv("C:\\Users\\GSA\\Workspace\\ALGORITMOS_IA\\CURSO_IA\\AMCLASS_METODOS_APRENDIZAGEM\\dados_bruto\\diabetesgravidez.csv")

# Mostra as cinco primeiras linhas da tabela.
# É uma primeira etapa de EDA: permite conferir os nomes
# das colunas e observar se os dados foram carregados corretamente.
base.head()
```

```python
# Lista com os nomes das características que serão usadas
# como entradas do modelo de IA.
# Em Machine Learning, estas colunas são normalmente chamadas:
# X, features, atributos ou variáveis independentes.
conj_caract = ['pregnant', # número de gestações
               'insulin',  # nível de insulina
               'bmi',      # índice de massa corporal
               'age',      # idade
               'glucose',  # glicose
               'bp',       # pressão arterial 
               'pedigree'  # histórico familiar/genético de diabetes
]

# Cria a variável "caract" contendo apenas as colunas escolhidas.
# caract será a matriz X do modelo:
# linhas = pacientes
# colunas = características clínicas.
caract = base[conj_caract] 

# Define a variável de saída, também chamada de alvo, target ou y.
# A coluna "label" deve conter a classe que se quer prever.
# Normalmente:
# 0 = não possui a condição
# 1 = possui a condição
alvo = base.label

# Importa a função que separa os dados em treino e teste
from sklearn.model_selection import train_test_split

# Divide os dados em quatro partes:
# caract_trein = características usadas para treinar o modelo
# caract_teste = características reservadas para testar o modelo
# alvo_trein   = respostas corretas do conjunto de treino
# alvo_teste   = respostas corretas do conjunto de teste
# test_size=0.3:
# reserva 30% dos dados para teste e usa 70% para treinamento.
# random_state=1:
# fixa a semente aleatória. Assim, ao executar novamente,
# a divisão será a mesma e o resultado poderá ser reproduzido.
caract_trein, caract_teste, alvo_trein, alvo_teste = train_test_split(caract, alvo, test_size=0.3, random_state=1)
```

```python
# Importa o modelo de Regressão Logística.
# Apesar do nome "regressão", este algoritmo é usado aqui
# para classificação binária: prever classe 0 ou classe 1.
from sklearn.linear_model import LogisticRegression

#Importa a métrica de acurácia.
# Acurácia = quantidade de previsões corretas / total de previsões.
from sklearn.metrics import accuracy_score

# Cria o objeto do modelo.
# solver='lbfgs' é o algoritmo numérico usado para encontrar
# os coeficientes da Regressão Logística.
logreg = LogisticRegression(solver='lbfgs')

# Treina o modelo.
# O algoritmo recebe:
# - caract_trein: dados de entrada X
# - alvo_trein: resposta conhecida y
# Durante o treinamento, o modelo encontra uma relação matemática
# entre glicose, IMC, idade, insulina etc. e a classe label.
logreg.fit(caract_trein,alvo_trein)

# Faz previsões para os dados de teste.
# Cada valor de predicoes será, por exemplo:
# 0 = classe negativa
# 1 = classe positiva
predicoes = logreg.predict(caract_teste)

# Compara os valores previstos pelo modelo com os valores reais.
# Quanto mais próxima de 1, maior a acurácia.
print("Logistic Regression Model Accuracy:",accuracy_score(alvo_teste, predicoes))
# Obrigatoriamente realizar a Tarefa 02 deste roteiro
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Separa treino e teste mantendo, aproximadamente,
# a mesma proporção de classes 0 e 1 em ambos os conjuntos.
X_trein, X_teste, y_trein, y_teste = train_test_split(
    caract,
    alvo,
    test_size=0.30,
    random_state=1,
    stratify=alvo
)

# Cria o padronizador.
scaler = StandardScaler()

# Aprende média e desvio padrão somente com os dados de treino.
X_trein_pad = scaler.fit_transform(X_trein)

# Aplica a mesma transformação aos dados de teste.
X_teste_pad = scaler.transform(X_teste)

# Cria e treina o modelo.
modelo = LogisticRegression(
    solver="lbfgs",
    max_iter=1000
)

modelo.fit(X_trein_pad, y_trein)

# Faz previsões.
y_pred = modelo.predict(X_teste_pad)

# Avalia o modelo.
print("Acurácia:", accuracy_score(y_teste, y_pred))
print("\nMatriz de confusão:")
print(confusion_matrix(y_teste, y_pred))

print("\nRelatório de classificação:")
print(classification_report(y_teste, y_pred))
```
